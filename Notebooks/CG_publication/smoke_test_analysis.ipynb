```python
import pandas as pd

smoke_test_path = "/home/stulcrad/master_thesis/Experiment_results/Smoke_Tests/CG"

# open all csv files in the smoke_test_path directory and create a dataframe with the results
import os
import glob

results = []
for file in glob.glob(os.path.join(smoke_test_path, "*.csv")):
    df = pd.read_csv(file)
    results.append(df)

display(pd.concat(results, ignore_index=True))
```
